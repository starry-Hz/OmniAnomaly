```python
import os
import numpy as np

pkl_files = ['omi-5_test_label.pkl']
for pkl_file in pkl_files:
    # 读取 .pkl 文件（假设是 NumPy 数组）
    data_path = os.path.join('processed', pkl_file)
    data = np.load(data_path, allow_pickle=True)  # 或者 pd.read_pickle(data_path)
    
    # 定义输出 .txt 文件路径
    txt_file = os.path.splitext(pkl_file)[0] + '.txt'
    txt_path = os.path.join('processed', txt_file)
    
    # 保存为 .txt 文件（默认空格分隔，可自定义格式）
    np.savetxt(txt_path, data, fmt='%s', delimiter=' ')  # delimiter 可以是逗号、制表符等
    
    print(f"已保存为 {txt_file}")
```

```python
def parse_txt_file(txt_file_path):
    """
    解析 .txt 文件，提取 start 和 end 范围
    """
    ranges = []
    with open(txt_file_path, 'r') as file:
        for line in file:
            # 去掉换行符
            line = line.strip()
            if not line:
                continue
            # 分割 start-end 和后面的内容
            range_part, _ = line.split(':', 1)
            start, end = map(int, range_part.split('-'))
            ranges.append((start, end))
    return ranges

def compare_results(txt_file_path, pkl_file_path):
    """
    对比 .txt 文件和 .pkl 文件的结果
    """
    # 解析 .txt 文件
    txt_ranges = parse_txt_file(txt_file_path)
    
    # 获取 .pkl 文件的结果
    pkl_result = find_consecutive_ranges(pkl_file_path)
    pkl_ranges = pkl_result[['start', 'end']].to_records(index=False).tolist()
    
    # 对比结果
    if txt_ranges != pkl_ranges:
        print("结果不一致！对比文件：", txt_file_path, pkl_file_path)
        print("txt 文件范围:", txt_ranges)
        print("pkl 文件范围:", pkl_ranges)
        print("-" * 100)

```

```python
import numpy as np
import pandas as pd
import os

def find_consecutive_ranges(file_path):
    """
    读取 .txt 文件，查找所有值为 1 的索引，并找出连续的范围（左闭右开）
    """
    with open(file_path, 'r') as file:
        data = file.readlines()

    # 将数据转换为 Pandas DataFrame
    df = pd.DataFrame(data)

    # 找到所有值为 1 的索引（行号从 1 开始）
    indices = df[df.iloc[:, 0].astype(int) == 1].index

    # 转换为 NumPy 数组
    indices = np.array(indices)

    # 如果 indices 为空，返回空 DataFrame
    if len(indices) == 0:
        return pd.DataFrame(columns=['start', 'end', 'length'])

    # 计算连续范围
    ranges = []
    start = indices[0]
    prev = start

    for i in range(1, len(indices)):
        if indices[i] != prev + 1:
            ranges.append((start, prev + 1))  # 使区间变为 [start, end)
            start = indices[i]
        prev = indices[i]
    ranges.append((start, prev + 1))  # 最后一段

    # 计算长度
    lengths = [end - start for start, end in ranges]  # 确保长度计算正确

    # 创建 DataFrame
    result = pd.DataFrame(ranges, columns=['start', 'end'])
    result['length'] = lengths

    return result

def parse_txt_file(txt_file_path):
    """
    解析 .txt 文件，提取 start 和 end 范围（左闭右开）
    """
    ranges = []
    with open(txt_file_path, 'r') as file:
        for line in file:
            line = line.strip()
            if not line:
                continue
            # 分割 start-end 和后面的内容
            range_part, _ = line.split(':', 1)
            start, end = map(int, range_part.split('-'))
            ranges.append((start, end))
    return ranges

def compare_results(txt_file_path, generated_txt_file_path):
    """
    对比 interpretation_label 和 test_label 目录下的 .txt 文件，**仅输出不一致的区间**
    """
    # 解析 interpretation_label 目录下的 txt 文件
    txt_ranges = parse_txt_file(txt_file_path)
    
    # 获取 test_label 目录下的 txt 文件处理结果
    generated_result = find_consecutive_ranges(generated_txt_file_path)
    generated_ranges = generated_result[['start', 'end']].to_records(index=False).tolist()
    
    # 计算不一致的区间
    missing_in_test_label = set(txt_ranges) - set(generated_ranges)  # interpretation_label 有但 test_label 没有
    extra_in_test_label = set(generated_ranges) - set(txt_ranges)  # test_label 有但 interpretation_label 没有
    
    # 只打印有差异的部分
    if missing_in_test_label or extra_in_test_label:
        txt_file_name = os.path.basename(txt_file_path)
        print(f"结果不一致！文件: {txt_file_name}")
        if missing_in_test_label:
            print("interpretation_label 多出的区间:", sorted(missing_in_test_label))
        if extra_in_test_label:
            print("test_label 多出的区间:", sorted(extra_in_test_label))
        print("-" * 100)

# 指定文件夹路径
interpretation_label_dir = 'ServerMachineDataset/interpretation_label'
test_label_dir = 'ServerMachineDataset/test_label'

# 遍历 interpretation_label 目录下的 machine- 开头的 .txt 文件
for txt_file_name in os.listdir(interpretation_label_dir):
    if txt_file_name.startswith('machine-') and txt_file_name.endswith('.txt'):
        txt_file_path = os.path.join(interpretation_label_dir, txt_file_name)
        
        # 构造 test_label 目录下的对应 txt 文件路径
        generated_txt_file_path = os.path.join(test_label_dir, txt_file_name)
        
        # 检查对应的 .txt 文件是否存在
        if os.path.exists(generated_txt_file_path):
            compare_results(txt_file_path, generated_txt_file_path)
        else:
            print(f"未找到对应的 .txt 文件: {generated_txt_file_path}")
```

```python
import os

# 指定路径
# path = r"OmniAnomaly\ServerMachineDataset\test_label"
path = './ServerMachineDataset/test_label'

# 初始化总计数器
total_counts = {'0': 0, '1': 0}

# 遍历目录下的所有.txt文件
for filename in os.listdir(path):
    if filename.endswith('.txt'):
        filepath = os.path.join(path, filename)
        with open(filepath, 'r') as file:
            # 初始化当前文件的计数器
            counts = {'0': 0, '1': 0}
            for line in file:
                line = line.strip()
                if line in ['0', '1']:
                    counts[line] += 1
                    total_counts[line] += 1
            # 打印当前文件的统计结果
            print(f"文件 {filename}: 0 的数量 = {counts['0']}, 1 的数量 = {counts['1']}")

# 打印总计结果
print(f"\n总计: 0 的数量 = {total_counts['0']}, 1 的数量 = {total_counts['1']}")
```
